## **1. Qué es un entorno virtual en Python y para qué sirve**

Un entorno virtual es una carpeta aislada que contiene su propia versión de Python y sus propias librerías instaladas. Cada proyecto puede tener su entorno sin afectar a los demás ni al Python general del computador.

**Para qué sirve**

* Evita conflictos entre versiones de librerías. Por ejemplo, un proyecto puede necesitar una versión antigua de una librería y otro proyecto una versión nueva.
* Mantiene limpio el sistema, porque todo lo que se instala queda dentro del entorno.
* Permite compartir un proyecto y que otra persona instale exactamente las mismas librerías.
* Permite probar librerías sin riesgo, ya que si algo sale mal se borra el entorno y se crea uno nuevo.

Se puede pensar en un entorno virtual como una caja de herramientas independiente para cada proyecto.

````python
import sys

print("Python en uso:", sys.executable)
print("Versión de Python:", sys.version)

# Si el prefijo actual es distinto al prefijo base, se está dentro de un entorno virtual
en_entorno_virtual = sys.prefix != sys.base_prefix
print("Dentro de un entorno virtual:", en_entorno_virtual)
````

## **2. Diferencia entre Anaconda y virtualenv**

Ambas herramientas sirven para crear entornos aislados, pero funcionan de forma distinta. Anaconda es una plataforma completa orientada a ciencia de datos, mientras que virtualenv es una herramienta pequeña que solo crea entornos de Python.

| Aspecto | Anaconda (conda) | virtualenv |
|---|---|---|
| Qué es | Distribución con gestor de paquetes y de entornos | Herramienta para crear entornos aislados |
| Gestor de paquetes | conda (también permite usar pip) | pip |
| Versión de Python | Puede instalar la versión de Python que se pida al crear el entorno | Usa un Python que ya esté instalado en el sistema |
| Tipo de paquetes | Python y también librerías externas (por ejemplo, de cálculo numérico) | Solo paquetes de Python |
| Tamaño | Pesado (varios GB). Miniconda es la versión ligera | Muy liviano |
| Ubicación de los entornos | Carpeta central de conda | Carpeta dentro del proyecto |
| Uso frecuente | Ciencia de datos e inteligencia artificial | Desarrollo general y aplicaciones web |

**Cuál elegir**

Si el trabajo se enfoca en análisis de datos o inteligencia artificial, conda suele ser más cómodo. Si se busca algo simple y rápido para un proyecto de programación general, virtualenv es suficiente.

## **5. Comentarios en Python**

Los comentarios son notas escritas para las personas que leen el código. Python las ignora al ejecutar. Sirven para explicar qué hace el código y para desactivar líneas temporalmente.

**Tipos de comentarios**

* Comentario de una línea: empieza con el símbolo `#`.
* Comentario al final de una línea de código: se escribe después del código con `#`.
* Comentario de varias líneas: Python no tiene una sintaxis especial, así que se escribe un `#` al inicio de cada línea.
* Docstring: texto entre comillas triples que documenta una función o un archivo. Técnicamente es una cadena de texto, pero se usa como documentación.

````python
# Este es un comentario de una línea
print("Hola")  # Este comentario va al final de una línea de código

# Este es un comentario
# de varias líneas, escrito
# con un símbolo numeral en cada una

# Desactivar una línea de código con un comentario:
# print("Esta línea no se ejecuta")

def saludar(nombre):
    '''Docstring: esta función imprime un saludo para la persona indicada.'''
    print("Hola,", nombre)

saludar("Ana")
print(saludar.__doc__)
````

## **6. La función print**

La función `print` muestra información en pantalla. Por defecto, cada llamada a `print` termina con un salto de línea, es decir, lo siguiente que se imprima aparece en una línea nueva.

**Parámetros útiles**

* `sep`: define el separador entre varios valores. Por defecto es un espacio.
* `end`: define lo que se escribe al final. Por defecto es un salto de línea. Si se cambia por una cadena vacía, el siguiente `print` continúa en la misma línea.

También se puede insertar un salto de línea dentro de un texto con `\n`.

````python
# print regular: cada llamada termina en una línea nueva
print("Primera línea")
print("Segunda línea")

# Varios valores separados por coma se imprimen con un espacio entre ellos
print("Hola", "mundo", 2024)

# Cambiar el separador
print("2024", "09", "23", sep="-")

# print sin salto de línea: el siguiente texto continúa en la misma línea
print("Cargando", end="")
print("...", end="")
print(" listo")

# Agregar saltos de línea dentro del texto con \n
print("Línea uno\nLínea dos\nLínea tres")

# print vacío: imprime una línea en blanco
print()

# Insertar variables dentro de un texto con f-string
nombre = "Ana"
edad = 20
print(f"Me llamo {nombre} y tengo {edad} años")
````

## **7. Variables: declaración y gramática**

Una variable es un nombre que guarda un valor en la memoria. En Python no se declara el tipo de la variable: basta con escribir el nombre, el signo igual y el valor. Python detecta el tipo por sí solo, y ese tipo puede cambiar si se le asigna otro valor.

**Gramática de una asignación**

```
nombre_de_la_variable = valor
```

**Reglas para los nombres de variables**

* Solo pueden contener letras, números y guion bajo (`_`).
* No pueden empezar con un número.
* No pueden tener espacios ni símbolos como `-`, `$` o `@`.
* Distinguen mayúsculas de minúsculas: `edad` y `Edad` son variables distintas.
* No pueden ser una palabra reservada del lenguaje.

Por convención, se usan letras minúsculas y guiones bajos para separar palabras, por ejemplo `nombre_completo`.

````python
# Declaración de variables
nombre_estudiante = "Ana"
edad = 20
promedio = 4.5
esta_matriculada = True

print(nombre_estudiante, edad, promedio, esta_matriculada)

# Asignación múltiple
x, y, z = 1, 2, 3
print(x, y, z)

# Asignar el mismo valor a varias variables
a = b = 0
print(a, b)

# Intercambiar el valor de dos variables
primero = "uno"
segundo = "dos"
primero, segundo = segundo, primero
print(primero, segundo)

# Python distingue mayúsculas y minúsculas
edad = 20
Edad = 30
print(edad, Edad)

# El tipo de una variable puede cambiar
dato = 10
print(dato, type(dato))
dato = "diez"
print(dato, type(dato))
````

## **8. Palabras reservadas e identificadores**

Las palabras reservadas son palabras que Python usa para su propia sintaxis, como `if`, `for` o `class`. Por eso no se pueden usar como nombres de variables, funciones ni ninguna otra cosa que se nombre en el código. Un identificador es cualquier nombre que elige el programador.

El módulo `keyword` permite consultar la lista de palabras reservadas de la versión de Python que se está usando, y comprobar si un nombre es válido.

````python
import keyword

# Lista de palabras reservadas de esta versión de Python
print(keyword.kwlist)
print("Cantidad de palabras reservadas:", len(keyword.kwlist))

# Comprobar si una palabra es reservada
print("'if' es reservada:", keyword.iskeyword("if"))
print("'nombre' es reservada:", keyword.iskeyword("nombre"))
````

````python
# Comprobar qué nombres se pueden usar como identificador
candidatos = ["nombre", "nombre_completo", "_privada", "edad2", "2edad",
              "mi-variable", "mi variable", "class", "total$", "None"]

for candidato in candidatos:
    es_valido = candidato.isidentifier() and not keyword.iskeyword(candidato)
    print(candidato, "->", es_valido)
````

## **9. Tipos de datos y la función type**

Los tipos de datos básicos de Python son los siguientes:

* `int`: números enteros, por ejemplo `7` o `-15`.
* `float`: números reales (con decimales), por ejemplo `3.14`.
* `str`: cadenas de texto, por ejemplo `"Hola"`.
* `bool`: valores booleanos, que solo pueden ser `True` o `False`.

La función `type()` indica el tipo de un valor o de una variable. También se puede convertir entre tipos con `int()`, `float()`, `str()` y `bool()`.

````python
entero = 7
real = 3.14
cadena = "Hola"
booleano = True

print(entero, type(entero))
print(real, type(real))
print(cadena, type(cadena))
print(booleano, type(booleano))

# El tipo del resultado depende de los tipos que se combinan
print(entero + real, type(entero + real))    # entero + real da un real
print(cadena + " mundo")                     # texto + texto une los textos
print(booleano + 1)                          # True vale 1 y False vale 0

# Conversión entre tipos
print(int("25") + 5)      # de texto a entero
print(float("3.5"))       # de texto a real
print(str(100) + "%")     # de número a texto
print(int(3.9))           # de real a entero (descarta los decimales)
print(bool(0), bool(7))   # 0 es False y cualquier otro número es True
````

## **10. Operadores aritméticos**

Los operadores aritméticos realizan cálculos con números enteros y reales.

| Operador | Nombre | Ejemplo | Resultado |
|---|---|---|---|
| `+` | Suma | `7 + 2` | 9 |
| `-` | Resta | `7 - 2` | 5 |
| `*` | Multiplicación | `7 * 2` | 14 |
| `/` | División (siempre da un real) | `7 / 2` | 3.5 |
| `//` | División entera (descarta los decimales) | `7 // 2` | 3 |
| `%` | Residuo o módulo | `7 % 2` | 1 |
| `**` | Potencia | `7 ** 2` | 49 |

Prueba cambiar los valores de `a` y `b` en las siguientes celdas para ver cómo cambian los resultados.

````python
# Operaciones con números enteros
a = 17
b = 5

print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicación:", a * b)
print("División:", a / b)
print("División entera:", a // b)
print("Residuo:", a % b)
print("Potencia:", a ** 2)
````

````python
# Operaciones con números reales
precio = 12.5
cantidad = 4
descuento = 0.1

subtotal = precio * cantidad
total = subtotal - subtotal * descuento
print("Subtotal:", subtotal)
print("Total con descuento:", total)

# Raíz cuadrada usando potencia
print("Raíz cuadrada de 16:", 16 ** 0.5)

# Los números reales pueden tener pequeños errores de precisión
print(0.1 + 0.2)
print(round(0.1 + 0.2, 2))
````

````python
# Usos frecuentes del residuo y la división entera
numero = 14
print("¿14 es par?", numero % 2 == 0)

segundos = 3725
horas = segundos // 3600
minutos = (segundos % 3600) // 60
print(f"{segundos} segundos son {horas} hora(s) y {minutos} minuto(s)")

# divmod entrega la división entera y el residuo al mismo tiempo
print(divmod(17, 5))
````

## **11. Cadenas: operadores, secuencias de escape e indexación**

Una cadena es un texto entre comillas simples o dobles. Se pueden usar algunos operadores con ellas, aunque no todos hacen lo mismo que con los números.

**Operadores con cadenas**

* `+` une dos cadenas (concatenación).
* `*` repite una cadena cierta cantidad de veces.
* `in` y `not in` comprueban si un texto está dentro de otro.
* `==`, `!=`, `<` y `>` comparan cadenas siguiendo el orden alfabético.
* `len()` entrega la cantidad de caracteres.

````python
nombre = "Python"
saludo = "Hola"

print(saludo + " " + nombre)       # concatenación
print(nombre * 3)                  # repetición
print("-" * 20)                    # línea decorativa
print("Py" in nombre)              # ¿está "Py" dentro de "Python"?
print("java" not in nombre)        # ¿"java" NO está dentro de "Python"?
print("casa" == "casa")            # igualdad
print("casa" == "Casa")            # distingue mayúsculas de minúsculas
print("ana" < "beto")              # orden alfabético
print(len(nombre))                 # cantidad de caracteres

# Con cadenas no se puede sumar un número directamente
edad = 20
print("Tengo " + str(edad) + " años")   # se convierte el número a texto primero
````

**Secuencias de escape**

Una secuencia de escape es una combinación que empieza con una barra invertida (`\`) y representa un carácter especial dentro de un texto.

| Secuencia | Significado |
|---|---|
| `\n` | Salto de línea |
| `\t` | Tabulación (espacio grande) |
| `\\` | Barra invertida |
| `\'` | Comilla simple |
| `\"` | Comilla doble |

Si se antepone la letra `r` a la cadena (cadena cruda), Python no interpreta las secuencias de escape. Esto es útil para rutas de archivos.

````python
print("Primera línea\nSegunda línea")
print("Nombre:\tAna")
print("Edad:\t20")
print("Ruta con barra invertida: C:\\Usuarios\\Ana")
print('Ella dijo: \'Hola\'')
print("Ella dijo: \"Hola\"")

# Cadena cruda: no interpreta las secuencias de escape
print(r"C:\nueva_carpeta\tarea")
````

**Indexación y segmentación**

Cada carácter de una cadena tiene una posición llamada índice. El primer carácter tiene el índice 0. También se puede contar desde el final usando índices negativos, donde `-1` es el último carácter.

Para obtener un fragmento se usa la segmentación (slicing) con la forma `[inicio:fin:paso]`. El índice de inicio se incluye, pero el de fin no.

```
 P   y   t   h   o   n
 0   1   2   3   4   5
-6  -5  -4  -3  -2  -1
```

````python
palabra = "Python"

print(palabra[0])        # primer carácter
print(palabra[5])        # último carácter
print(palabra[-1])       # último carácter, contando desde el final
print(palabra[-2])       # penúltimo carácter

print(palabra[0:2])      # desde el índice 0 hasta el 1
print(palabra[2:])       # desde el índice 2 hasta el final
print(palabra[:3])       # desde el inicio hasta el índice 2
print(palabra[::2])      # de dos en dos
print(palabra[::-1])     # la cadena al revés

# Las cadenas no se pueden modificar por índice
try:
    palabra[0] = "J"
except TypeError as error:
    print("Error:", error)

# Para "cambiar" una cadena se crea una nueva
nueva_palabra = "J" + palabra[1:]
print(nueva_palabra)
````

**Métodos comunes de cadenas**

Además de los operadores, las cadenas tienen métodos, que son funciones que se escriben después de un punto.

````python
frase = "  Hola Mundo desde Python  "

print(frase.upper())              # todo en mayúsculas
print(frase.lower())              # todo en minúsculas
print(frase.strip())              # quita los espacios de los bordes
print(frase.replace("Mundo", "Colab"))
print(frase.strip().split(" "))   # separa el texto en una lista de palabras
print(frase.count("o"))           # cuántas veces aparece la letra o
print(frase.find("Python"))       # posición donde empieza "Python"
````

## **12. Booleanos, comparaciones y operadores lógicos**

Un valor booleano solo puede ser `True` (verdadero) o `False` (falso). Se obtienen al comparar valores y se combinan con operadores lógicos.

**Operadores de comparación**

Comparan dos valores y devuelven un booleano: `==` (igual), `!=` (diferente), `<` (menor), `>` (mayor), `<=` (menor o igual) y `>=` (mayor o igual).

**Operadores lógicos**

* `and`: es verdadero solo si ambas condiciones son verdaderas.
* `or`: es verdadero si al menos una condición es verdadera.
* `not`: invierte el valor, lo verdadero pasa a falso y viceversa.

````python
edad = 20
nota = 4.2

# Comparaciones
print(edad == 20)
print(edad != 18)
print(edad < 18)
print(edad >= 18)
print(3 < edad < 30)     # comparación encadenada

# Operadores lógicos
es_mayor = edad >= 18
aprobo = nota >= 3.0

print(es_mayor and aprobo)
print(es_mayor or aprobo)
print(not aprobo)

# Tabla de verdad de and y or
print("A", "B", "A and B", "A or B", sep=" | ")
for valor_a in (True, False):
    for valor_b in (True, False):
        print(valor_a, valor_b, valor_a and valor_b, valor_a or valor_b, sep=" | ")
````

**Valores que se consideran verdaderos o falsos**

Al convertir un valor a booleano con `bool()`, Python considera falsos el número 0, la cadena vacía, las colecciones vacías y `None`. Todo lo demás se considera verdadero.

````python
print(bool(0), bool(1), bool(-5))
print(bool(""), bool("hola"))
print(bool([]), bool([1, 2]))
print(bool(None))

# Los operadores lógicos también funcionan con estos valores
print("" or "valor por defecto")
print("texto" and "resultado")
````

## **13. Operadores bit a bit**

Los operadores bit a bit trabajan con la representación binaria de los números enteros, es decir, con ceros y unos. La función `bin()` permite ver un número en binario.

| Operador | Nombre | Qué hace |
|---|---|---|
| `&` | AND | Da 1 solo si ambos bits son 1 |
| `\|` | OR | Da 1 si al menos uno de los bits es 1 |
| `^` | XOR | Da 1 si los bits son diferentes |
| `~` | NOT | Invierte todos los bits |
| `<<` | Desplazamiento a la izquierda | Mueve los bits a la izquierda (equivale a multiplicar por 2) |
| `>>` | Desplazamiento a la derecha | Mueve los bits a la derecha (equivale a dividir entre 2) |

````python
a = 5    # en binario: 101
b = 3    # en binario: 011

print("a en binario:", bin(a))
print("b en binario:", bin(b))

print("a & b:", a & b, "->", bin(a & b))
print("a | b:", a | b, "->", bin(a | b))
print("a ^ b:", a ^ b, "->", bin(a ^ b))
print("~a:", ~a)             # ~x es igual a -(x + 1)
print("a << 1:", a << 1)     # 5 * 2
print("a << 2:", a << 2)     # 5 * 4
print("a >> 1:", a >> 1)     # 5 // 2
````

## **14. Operadores de identidad y pertenencia**

**Identidad**

`is` comprueba si dos nombres apuntan exactamente al mismo objeto en memoria, mientras que `==` comprueba si dos valores son iguales. `is not` es la negación. Se usa mucho para comparar con `None`.

**Pertenencia**

`in` comprueba si un elemento está dentro de una colección o de un texto, y `not in` comprueba lo contrario.

````python
lista_a = [1, 2, 3]
lista_b = [1, 2, 3]
lista_c = lista_a

print(lista_a == lista_b)       # tienen el mismo contenido
print(lista_a is lista_b)       # pero son objetos distintos
print(lista_a is lista_c)       # lista_c es el mismo objeto que lista_a
print(lista_a is not lista_b)

resultado = None
print(resultado is None)

# Pertenencia
frutas = ["manzana", "pera", "uva"]
print("pera" in frutas)
print("mango" not in frutas)
print("th" in "Python")
print(2 in [1, 2, 3])

# En un diccionario, in busca entre las claves
estudiante = {"nombre": "Ana", "edad": 20}
print("nombre" in estudiante)
print("Ana" in estudiante)
````

## **15. Operadores de asignación compuestos**

Un operador de asignación compuesto combina una operación con la asignación. Sirve para actualizar una variable a partir de su propio valor de forma abreviada. Por ejemplo, `x += 3` es lo mismo que `x = x + 3`.

| Operador | Equivale a |
|---|---|
| `x += 3` | `x = x + 3` |
| `x -= 3` | `x = x - 3` |
| `x *= 3` | `x = x * 3` |
| `x /= 3` | `x = x / 3` |
| `x //= 3` | `x = x // 3` |
| `x %= 3` | `x = x % 3` |
| `x **= 3` | `x = x ** 3` |

````python
x = 10
print("Valor inicial:", x)

x += 5
print("x += 5  ->", x)

x -= 3
print("x -= 3  ->", x)

x *= 2
print("x *= 2  ->", x)

x /= 4
print("x /= 4  ->", x)

x = 17
x //= 5
print("x //= 5 ->", x)

x = 17
x %= 5
print("x %= 5  ->", x)

x = 2
x **= 5
print("x **= 5 ->", x)

# También funciona con cadenas
mensaje = "Hola"
mensaje += " mundo"
print(mensaje)

# Un contador típico
contador = 0
for i in range(5):
    contador += 1
print("Contador:", contador)
````

## **16. Precedencia de operadores**

La precedencia define qué operación se resuelve primero cuando una expresión tiene varios operadores. Si dos operadores tienen la misma prioridad, casi siempre se resuelven de izquierda a derecha. Los paréntesis permiten forzar el orden deseado.

La siguiente tabla muestra los operadores de mayor a menor prioridad. Cada fila es un nivel de prioridad.

| Operador | Explicación |
|---|---|
| `()` | operador de paréntesis |
| `**` | operador de potencia |
| `~`, `+`, `-` | operador monádico |
| `*`, `/`, `%`, `//` | multiplicación, división, operador división resto, operador división entero |
| `+`, `-` | suma, resta |
| `>>`, `<<` | operador de movimiento bit a bit |
| `&` | operador AND |
| `^`, &#124; | operador XOR bit a bit, operador OR bit |
| `<=`, `<`, `>`, `>=` | operador de comparación |
| `==`, `!=` | operador igual, operador diferente |
| `=`, `%=`, `/=`, `//=`, `-=`, `+=`, `*=`, `**=` | operador de asignación, operador de asignación compleja |
| `is`, `is not` | operador de identidad |
| `in`, `not in` | operador de prueba de afiliación |
| `not`, `or`, `and` | operador lógico |

**Nota**

En Python, los operadores de comparación, identidad y pertenencia comparten en la práctica el mismo nivel de prioridad. Entre los operadores lógicos se evalúa primero `not`, luego `and` y al final `or`.

````python
print(2 + 3 * 4)          # la multiplicación va antes que la suma
print((2 + 3) * 4)        # los paréntesis cambian el orden
print(2 ** 3 ** 2)        # la potencia se resuelve de derecha a izquierda: 2 ** 9
print(-2 ** 2)            # primero la potencia y luego el signo
print((-2) ** 2)          # con paréntesis el signo se aplica primero
print(10 - 4 - 3)         # misma prioridad: de izquierda a derecha
print(100 / 10 * 2)       # de izquierda a derecha
print(1 + 2 << 2)         # primero la suma (3) y luego el desplazamiento
print(7 + 3 > 5 and 2 * 2 == 4)   # primero los cálculos, luego comparaciones y por último and

# Orden entre operadores lógicos: not, luego and, luego or
print(True or False and False)    # and se resuelve primero: True or False
print(not True or True)           # not se resuelve primero: False or True
print(not (True or True))         # con paréntesis cambia el resultado
````

## **17. Estructuras de datos: listas, tuplas, diccionarios y conjuntos**

Las estructuras de datos permiten guardar varios valores en una sola variable. Las cuatro más importantes de Python son:

| Estructura | Símbolo | Ordenada | Se puede modificar | Permite repetidos |
|---|---|---|---|---|
| Lista (`list`) | `[ ]` | Sí | Sí | Sí |
| Tupla (`tuple`) | `( )` | Sí | No | Sí |
| Diccionario (`dict`) | `{clave: valor}` | Sí (por orden de inserción) | Sí | Claves únicas |
| Conjunto (`set`) | `{ }` | No | Sí | No |

**Listas**

Una lista guarda elementos en orden y se puede modificar: agregar, quitar o cambiar elementos. Se accede a ellos por índice, igual que en las cadenas.

````python
frutas = ["manzana", "pera", "uva"]
print(frutas)
print(frutas[0], frutas[-1])
print(len(frutas))

frutas.append("mango")          # agregar al final
frutas.insert(1, "fresa")       # insertar en una posición
print(frutas)

frutas.remove("pera")           # quitar por valor
ultimo = frutas.pop()           # quitar y devolver el último
print("Se quitó:", ultimo)
print(frutas)

frutas[0] = "kiwi"              # modificar un elemento
print(frutas)

numeros = [5, 2, 9, 1]
numeros.sort()
print(numeros, max(numeros), min(numeros), sum(numeros))

# Recorrer una lista
for fruta in frutas:
    print("Fruta:", fruta)
````

**Tuplas**

Una tupla es parecida a una lista, pero no se puede modificar una vez creada. Es útil para datos que no deben cambiar, como coordenadas o fechas.

````python
coordenadas = (10, 20)
print(coordenadas, type(coordenadas))
print(coordenadas[0])

# Desempaquetado: repartir los valores en variables
x, y = coordenadas
print("x =", x, "y =", y)

# Las tuplas no se pueden modificar
try:
    coordenadas[0] = 99
except TypeError as error:
    print("Error:", error)

# Una tupla de un solo elemento necesita una coma final
un_elemento = (5,)
print(type(un_elemento))

# Una tupla se puede convertir en lista y viceversa
lista_temporal = list(coordenadas)
lista_temporal.append(30)
print(tuple(lista_temporal))
````

**Diccionarios**

Un diccionario guarda pares de clave y valor. En lugar de acceder por posición, se accede por la clave, lo que lo hace ideal para representar datos con nombre, como los de una persona.

````python
estudiante = {
    "nombre": "Ana",
    "edad": 20,
    "carrera": "Sistemas"
}

print(estudiante["nombre"])
print(estudiante.get("telefono", "No registrado"))   # get evita errores si la clave no existe

estudiante["edad"] = 21               # modificar un valor
estudiante["ciudad"] = "Panamá"       # agregar un par nuevo
print(estudiante)

del estudiante["carrera"]             # eliminar un par
print(estudiante)

print(estudiante.keys())              # las claves
print(estudiante.values())            # los valores
print(estudiante.items())             # los pares

for clave, valor in estudiante.items():
    print(clave, "->", valor)
````

**Conjuntos**

Un conjunto guarda elementos únicos, sin orden. Elimina los repetidos automáticamente y permite hacer operaciones de conjuntos como unión, intersección y diferencia.

````python
numeros = {1, 2, 2, 3, 3, 3}
print(numeros)                        # los repetidos desaparecen

grupo_a = {"Ana", "Luis", "Marta"}
grupo_b = {"Luis", "Pedro", "Marta"}

print("Unión:", grupo_a | grupo_b)
print("Intersección:", grupo_a & grupo_b)
print("Diferencia:", grupo_a - grupo_b)

grupo_a.add("Sofía")
print(grupo_a)

# Truco frecuente: quitar repetidos de una lista
con_repetidos = [1, 1, 2, 3, 3]
print(list(set(con_repetidos)))
````

## **18. Errores de tiempo de ejecución**

Un error de tiempo de ejecución (runtime error) ocurre mientras el programa ya se está ejecutando. El código está bien escrito, pero al llegar a cierta línea sucede algo que Python no puede realizar, como dividir entre cero, usar una variable que no existe o convertir un texto que no es un número.

**Diferencia con otros tipos de errores**

* Error de sintaxis: el código está mal escrito y Python ni siquiera empieza a ejecutarlo. Ejemplo: olvidar unos paréntesis.
* Error de tiempo de ejecución: el programa empieza, pero se detiene al llegar a la línea con el problema.
* Error lógico: el programa funciona sin avisar, pero el resultado es incorrecto porque la lógica está mal planteada.

Cuando ocurre un error, Python muestra su tipo y un mensaje. Para que el programa no se detenga, se puede capturar el error con `try` y `except`. En la siguiente celda todos los errores están controlados para que puedas ver sus nombres y mensajes.

````python
# ZeroDivisionError: dividir entre cero
try:
    resultado = 10 / 0
except Exception as error:
    print(type(error).__name__, "->", error)

# NameError: usar una variable que no existe
try:
    print(variable_inexistente)
except Exception as error:
    print(type(error).__name__, "->", error)

# TypeError: mezclar tipos que no se pueden operar
try:
    resultado = "edad: " + 20
except Exception as error:
    print(type(error).__name__, "->", error)

# ValueError: el valor tiene el tipo correcto pero no es válido
try:
    numero = int("abc")
except Exception as error:
    print(type(error).__name__, "->", error)

# IndexError: usar un índice fuera de rango
try:
    lista = [1, 2, 3]
    print(lista[10])
except Exception as error:
    print(type(error).__name__, "->", error)

# KeyError: pedir una clave que no existe en un diccionario
try:
    datos = {"nombre": "Ana"}
    print(datos["edad"])
except Exception as error:
    print(type(error).__name__, "->", error)
````

* `except`: se ejecuta solo si ocurre un error.
* `else`: se ejecuta solo si no ocurrió ningún error.
* `finally`: se ejecuta siempre, haya o no error.

## **19. Notación Big-O**

Big-O es una forma de describir qué tan rápido crece el trabajo que hace un programa cuando aumenta la cantidad de datos. No mide segundos, sino cuántas operaciones se necesitan en función del tamaño de los datos, al que se llama `n`.

Sirve para comparar formas de resolver un mismo problema. Un programa puede funcionar bien con 10 datos y volverse lentísimo con un millón, y Big-O ayuda a prever eso. Se suele analizar el peor caso y se ignoran los detalles pequeños, porque lo que importa es el crecimiento.

**Órdenes de crecimiento más comunes**

| Notación | Nombre | Qué significa | Ejemplo |
|---|---|---|---|
| O(1) | Constante | El trabajo no cambia aunque haya más datos | Acceder a un elemento de una lista por su índice |
| O(log n) | Logarítmico | El trabajo crece muy poco al duplicar los datos | Búsqueda binaria en una lista ordenada |
| O(n) | Lineal | El trabajo crece igual que los datos | Recorrer una lista completa |
| O(n^2) | Cuadrático | Si los datos se duplican, el trabajo se cuadruplica | Un bucle dentro de otro bucle |

**Contando operaciones**

En la siguiente celda cada función cuenta cuántos pasos da. Observa cómo cambia el número de pasos cuando `n` crece de 10 a 100 y a 1000.

````python
def pasos_constante(n):
    # Siempre da un solo paso, sin importar n
    return 1

def pasos_logaritmico(n):
    # Divide n entre 2 hasta llegar a 1
    contador = 0
    while n > 1:
        n = n // 2
        contador += 1
    return contador

def pasos_lineal(n):
    # Un paso por cada elemento
    contador = 0
    for i in range(n):
        contador += 1
    return contador

def pasos_cuadratico(n):
    # Un bucle dentro de otro: n por n pasos
    contador = 0
    for i in range(n):
        for j in range(n):
            contador += 1
    return contador

print("n", "O(1)", "O(log n)", "O(n)", "O(n^2)", sep=" | ")
for n in [10, 100, 1000]:
    print(n, pasos_constante(n), pasos_logaritmico(n), pasos_lineal(n), pasos_cuadratico(n), sep=" | ")
````

**Por qué sucede esto**

* En O(1) el resultado no depende de `n`, por eso siempre es un paso.
* En O(log n) el problema se reduce a la mitad en cada paso, así que incluso con 1000 datos solo se necesitan unos 10 pasos.
* En O(n) hay un paso por dato: 10 datos son 10 pasos y 1000 datos son 1000 pasos.
* En O(n^2) hay un bucle dentro de otro: con 1000 datos son un millón de pasos. Por eso los bucles anidados se vuelven lentos rápidamente.

**Ejemplo real: buscar un número en una lista**

La búsqueda lineal revisa elemento por elemento (O(n)). La búsqueda binaria funciona solo en listas ordenadas: mira el elemento del medio y descarta la mitad de la lista en cada paso (O(log n)).

````python
def busqueda_lineal(lista, objetivo):
    pasos = 0
    for elemento in lista:
        pasos += 1
        if elemento == objetivo:
            break
    return pasos

def busqueda_binaria(lista, objetivo):
    pasos = 0
    inicio = 0
    fin = len(lista) - 1
    while inicio <= fin:
        pasos += 1
        medio = (inicio + fin) // 2
        if lista[medio] == objetivo:
            break
        elif lista[medio] < objetivo:
            inicio = medio + 1
        else:
            fin = medio - 1
    return pasos

numeros = list(range(1, 100001))     # lista ordenada del 1 al 100000
objetivo = 100000                    # el peor caso: el último elemento

print("Pasos con búsqueda lineal:", busqueda_lineal(numeros, objetivo))
print("Pasos con búsqueda binaria:", busqueda_binaria(numeros, objetivo))
````

**Ejemplo con tiempo real: lista contra conjunto**

Comprobar si un elemento está en una lista con `in` es O(n), porque Python puede tener que revisar todos los elementos. En un conjunto es O(1) en la práctica, porque usa una tabla interna que permite ir directo al elemento. La diferencia se nota al medir el tiempo.

````python
import time

cantidad = 200000
como_lista = list(range(cantidad))
como_conjunto = set(como_lista)
buscado = -1          # no existe, así que la lista tiene que revisar todo
repeticiones = 200

inicio = time.perf_counter()
for i in range(repeticiones):
    buscado in como_lista
tiempo_lista = time.perf_counter() - inicio

inicio = time.perf_counter()
for i in range(repeticiones):
    buscado in como_conjunto
tiempo_conjunto = time.perf_counter() - inicio

print("Tiempo con lista (O(n)):", round(tiempo_lista, 4), "segundos")
print("Tiempo con conjunto (O(1)):", round(tiempo_conjunto, 6), "segundos")
print("El conjunto fue aproximadamente", int(tiempo_lista / tiempo_conjunto), "veces más rápido")
````

## **20. Condicional if**

El condicional `if` ejecuta un bloque de código solo si una condición es verdadera. La condición es una expresión que da como resultado `True` o `False`. El bloque que depende del `if` se escribe con sangría (espacios al inicio de la línea), y solo se ejecuta cuando la condición se cumple.

## **22. Condicional if elif else**

`elif` significa "si no, y si..." y permite revisar varias condiciones en orden. Python revisa cada condición de arriba hacia abajo y ejecuta solo el primer bloque cuya condición sea verdadera. Si ninguna se cumple, se ejecuta el `else`, que es opcional.

```
if condicion_1:
    instrucción si la condición 1 es verdadera

```

## **23. La sangría y cómo cambia los resultados**

En Python la sangría no es un adorno: define qué instrucciones pertenecen a cada bloque. Todas las líneas con la misma sangría, justo debajo de un `if`, `elif` o `else`, forman parte de ese bloque. En cuanto una línea vuelve al nivel de sangría anterior, Python entiende que el bloque terminó.

Por eso, mover una línea unos espacios a la izquierda o a la derecha puede cambiar por completo el resultado del programa, aunque el código se vea casi igual.

````python
edad = 20
tiene_carnet = False

# Caso 1: la segunda instrucción está DENTRO del if (misma sangría que la primera)
if edad >= 18:
    print("Es mayor de edad")
    if tiene_carnet:
        print("Puede conducir")

print("----")

# Caso 2: la segunda instrucción está FUERA del if (sin sangría), aunque la condición sea la misma
if edad >= 18:
    print("Es mayor de edad")
if tiene_carnet:
    print("Puede conducir")
````

En el primer caso, el segundo `if` está dentro del primero (anidado), así que solo se revisa si la persona ya es mayor de edad. En el segundo caso, el segundo `if` es independiente y se revisa siempre, sin importar el resultado del primero. Con `tiene_carnet = False` ninguno imprime "Puede conducir", pero la lógica interna es distinta, y eso se nota si se cambian los valores.

````python
def revisar_con_anidado(edad, tiene_carnet):
    if edad >= 18:
        print("Es mayor de edad")
        if tiene_carnet:
            print("Puede conducir")

def revisar_sin_anidado(edad, tiene_carnet):
    if edad >= 18:
        print("Es mayor de edad")
    if tiene_carnet:
        print("Puede conducir")

print("Con anidado, edad=15, tiene_carnet=True:")
revisar_con_anidado(15, True)   # no imprime nada, porque no es mayor de edad

print("\nSin anidado, edad=15, tiene_carnet=True:")
revisar_sin_anidado(15, True)   # sí imprime "Puede conducir", aunque sea menor de edad
````

El ejemplo anterior muestra el riesgo de una sangría mal puesta: con `tiene_carnet = True` y una persona menor de edad, la versión sin anidar deja "conducir" como verdadero de todas formas, porque ese `if` ya no depende del primero. Por eso conviene revisar con cuidado qué instrucciones quedan dentro y cuáles quedan fuera de cada bloque.
