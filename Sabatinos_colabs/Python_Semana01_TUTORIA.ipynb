## Ejemplo de Base de Datos

A continuación, se muestra un ejemplo de cómo podría verse una base de datos de estudiantes utilizando listas y diccionarios:

Desarrolle el código en la siguiente lista.

````python
base_de_datos = [
    {"nombre": "Juan", "id": 1, "edad": 20},
    {"nombre": "María", "id": 2, "edad": 22},
    {"nombre": "Carlos", "id": 3, "edad": 21}
]

while True:
  print("\n--- BASE DE DATOS DE ESTUDIANTES ---")
  print("1. Agregar un nuevo estudiante.")
  print("2. Mostrar la lista de estudiantes.")
  print("3. Buscar un estudiante.")
  print("4. Actualizar la información.")
  print("5. Eliminar un estudiante.")
  print("6. Salir.")

  opcion = input("Ingrese una opción: ")

  if opcion == "1":
    nombre = input("Ingrese el nombre del estudiante: ")
    id = int(input("Ingrese el ID del estudiante: "))
    edad = int(input("Ingrese la edad del estudiante: "))

    existe = False

    for estudiante in base_de_datos:
      if estudiante["id"] == id:
        existe = True
        break

    if existe:
      print("El estudiante ya existe.")
    else:
      nuevo_estudiante = {"nombre": nombre, "id": id, "edad": edad}
      base_de_datos.append(nuevo_estudiante)
      print("Estudiante agregado correctamente.")

  elif opcion == "2":
    if len(base_de_datos) == 0:
      print("No hay estudiantes en la base de datos.")
    else:
      print("\n--- LISTA DE ESTUDIANTES ---")
      for estudiante in base_de_datos:
        print(estudiante)

  elif opcion == "3":
    id_buscar = int(input("Ingrese el ID del estudiante a buscar: "))
    encontrado = False
    for est in base_de_datos:
      if est["id"] == id_buscar:
        print(f"Encontrado -> Nombre: {est['nombre']}, Edad: {est['edad']}")
        encontrado = True
        break

      if not encontrado:
        print("Estudiante no encontrado.")

  elif opcion == "4":
    id_buscar = int(input("Ingrese el ID del estudiante a actualizar: "))
    encontrado = False

    for estudiante in base_de_datos:
        if estudiante["id"] == id_buscar:
            print(f"Estudiante encontrado: {estudiante['nombre']} (Edad: {estudiante['edad']})")
            nuevo_nombre = input("Ingrese el nuevo nombre: ")
            nueva_edad = int(input("Ingrese la nueva edad: "))
            estudiante["nombre"] = nuevo_nombre
            estudiante["edad"] = nueva_edad

            print("Información actualizada correctamente.")
            encontrado = True
            break

    if not encontrado:
        print("El estudiante con ese ID no existe en la base de datos.")

  elif opcion == "5":
    id_eliminar = int(input("Ingrese el ID del estudiante a eliminar: "))
    encontrado = False

    for est in base_de_datos:
      if est["id"] == id_eliminar:
        base_de_datos.remove(est)
        print("Estudiante eliminado correctamente.")
        encontrado = True
        break

      if not encontrado:
        print("Estudiante no encontrado.")

  elif opcion == "6":
    print("\n Hasta luego.")
    break

  else:
    print("Opción inválida. Por favor, ingrese una opción válida.")
````

# **Extras**

````python
import time

pedido_1 = {
    "numero": 1,
    "pizza": "Pepperoni",
    "ingredientes": ["queso", "pepperoni", "salsa"]
}

pedido_2 = {
    "numero": 2,
    "pizza": "Anchoas",
    "ingredientes": ["queso", "anchoas", "salsa", "aceitunas"]
}

pedido_3 = {
    "numero": 3,
    "pizza": "Hawaiana",
    "ingredientes": ["queso", "jamón", "piña", "salsa"]
}

pedido_4 = {
    "numero": 4,
    "pizza": "Vegetariana",
    "ingredientes": ["queso", "tomate", "champiñones", "pimiento"]
}

pendientes = []
preparando = []
entregadas = []

nuevos_pedidos = [pedido_1, pedido_2, pedido_3, pedido_4]

for pedido in nuevos_pedidos:
    pendientes.append(pedido)

    print("\n¡Nuevo pedido recibido!")
    print("Pedido:", pedido["numero"])
    print("Pizza:", pedido["pizza"])
    print("Ingredientes:", pedido["ingredientes"])
    print("Pedidos en espera:", len(pendientes))

    time.sleep(10)

# Preparar los pedidos en orden de llegada
while len(pendientes) > 0:

    # Tomar el primer pedido de la fila
    pedido_actual = pendientes.pop(0)

    # Pasar el pedido a preparación
    preparando.append(pedido_actual)

    print("\n------------------------------")
    print("Preparando pedido:", pedido_actual["numero"])
    print("Pizza:", pedido_actual["pizza"])
    print("Ingredientes necesarios:")

    for ingrediente in pedido_actual["ingredientes"]:
        print("-", ingrediente)

    print("Pedidos pendientes:", len(pendientes))
    print("Pedidos preparando:", len(preparando))
    print("Pedidos entregados:", len(entregadas))

    time.sleep(10)

    # La pizza termina de prepararse
    preparando.remove(pedido_actual)
    entregadas.append(pedido_actual)

    print("\n¡Pizza lista para entregar!")
    print("Pedido:", pedido_actual["numero"])
    print("Pizza:", pedido_actual["pizza"])

    print("Pedidos pendientes:", len(pendientes))
    print("Pedidos preparando:", len(preparando))
    print("Pedidos entregados:", len(entregadas))

    time.sleep(10)

print("\n========== RESUMEN FINAL ==========")
print("Total de pedidos recibidos:", len(nuevos_pedidos))
print("Pedidos pendientes:", len(pendientes))
print("Pedidos preparando:", len(preparando))
print("Pedidos entregados:", len(entregadas))

print("\nPizzas entregadas:")

for pedido in entregadas:
    print("Pedido", pedido["numero"], "-", pedido["pizza"])
````
