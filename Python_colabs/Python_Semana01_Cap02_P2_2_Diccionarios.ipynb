### 14.1 Modificación Segura: setdefault() y update()

**Problema que resuelve:** cargar o ampliar datos sin pisar por accidente lo que ya existe, y modificar muchos campos de una sola vez.

- `d.setdefault(clave, defecto)` devuelve el valor de `clave` si existe. Si no existe, la crea con `defecto` y devuelve ese valor. **Nunca sobrescribe** un valor previo.
- `d.update(otro)` inserta o modifica varios pares a la vez. Si la clave ya existe, **sí** reemplaza su valor.
- `update` acepta otro diccionario, una lista de pares `(clave, valor)` o argumentos con nombre: `d.update(a=1, b=2)`.

Regla práctica: `setdefault` para "inicializar si falta", `update` para "aplicar estos cambios".

```python
# Ejemplo 1 (General): setdefault y update
config = {"tema":"oscuro", "idioma": "es"}

devuelto = config.setdefault("tema", "claro")
print("tema:", devuelto)

config.setdefault

print("Tras setdefault:", config)

config.update(idioma="en", tema="claro")
print("Tras update:", config)
```

### 14.2 Consulta Segura y Limpieza: get() y clear()

**Problema que resuelve:** consultar claves que quizá no existan sin que el programa se detenga, y vaciar un diccionario cuando sus datos ya no sirven.

- `d[clave]` lanza `KeyError` si la clave no existe. `d.get(clave, valor_defecto)` devuelve el valor por defecto (o `None` si no se indica).
- `d.get()` **no modifica** el diccionario; `setdefault` sí.
- `d.clear()` elimina **todos** los pares, pero conserva el mismo objeto (otras variables que lo referencian lo verán vacío). Es distinto de `d = {}`, que crea un diccionario nuevo.

```python
# Ejemplo 1 (General): get y clear
precios = {"pan": 1.20, "leche": 2.50}

print(precios.get("pan"))
print("-" * 40)
print(precios.get("queso"))                 # None, sin error
print("-" * 40)
print(precios.get("queso", "no existe"))   #valor por defecto

try:
    print(precios["queso"])
except KeyError as error:
    print("KeyError esperado:", error)

print("get no modifica el diccionario:", precios)

# clear vacía el mismo objeto; una segunda variable lo ve vacío
alias = precios
precios.clear()
print("Tras clear:", precios, "| alias:", alias)

# Diferencia con reasignar
a = {"x": 1}
b = a
a = {}
print("Reasignar no afecta a b:", a, b)
```

### 15.1 Conversión y Mapeo Avanzado: fromkeys() y defaultdict

**Problema que resuelve:** crear diccionarios con una estructura inicial uniforme y acumular datos sin verificar si la clave ya existe.

- `dict.fromkeys(secuencia, valor)` crea un diccionario con todas las claves de la secuencia y el **mismo** valor inicial (por defecto `None`).
- **Trampa:** si el valor es mutable (por ejemplo `[]`), todas las claves comparten **la misma lista**.
- `collections.defaultdict(tipo)` crea automáticamente un valor inicial (`int` da `0`, `list` da `[]`) cuando se consulta una clave inexistente, sin lanzar `KeyError`.
- `defaultdict(int)` sirve para contar; `defaultdict(list)` para agrupar.

```python
# Ejemplo 1 (General): fromkeys y defaultdict
from collections import defaultdict

contadores = dict.fromkeys(["aya", "bya", "cya"], 10)
print("fromkeys con 10:", contadores)
print("fromkeys sin valor:", dict.fromkeys(["x", "y"],5))

print("//" * 40)

# Trampa del valor mutable compartido
comparten = dict.fromkeys(["a", "b"], [])
comparten["a"].append(1)
comparten["b"].append(2)
print("Lista compartida (error clásico):", comparten)

# Forma correcta: una lista distinta por clave
correcto = {}
for clave in ["a", "b"]:
    correcto[clave] = []
correcto["a"].append(1)
correcto["b"].append(2)
print("Listas independientes:", correcto)

# defaultdict(int): contar
conteo = defaultdict(int)
for letra in "banana":
    conteo[letra] += 1
print("Conteo de letras:", dict(conteo))
```

### 15.2 Recorrido de Estructuras Anidadas: .items(), .keys(), .values()

**Problema que resuelve:** leer y resumir datos cuando los valores de un diccionario son a su vez diccionarios o colecciones.

- `.items()` entrega pares `(clave, valor)`; si el valor es otro diccionario, se vuelve a llamar `.items()` en un segundo bucle.
- `.keys()` sirve para listar los niveles superiores y `.values()` para operar sobre los datos sin necesitar las claves (por ejemplo `sum(d.values())`).
- Regla de legibilidad: usa nombres de variable que indiquen el nivel (`categoria`, `producto`, `valor`).

```python
# Ejemplo 1 (General): recorrido de diccionarios anidados
notas = {
    "Matemáticas": {"Ana": 90, "Luis": 75},
    "Historia": {"Ana": 85, "Luis": 95, "Maria": 70},
}

print("Materias:", list(notas.keys()))

for materia, alumnos in notas.items():
    print(materia)
    for alumno, nota in alumnos.items():
        print(f"   {alumno}: {nota}")

# values() para resumir sin necesitar las claves
for materia, alumnos in notas.items():
    promedio = sum(alumnos.values()) / len(alumnos)
    print(f"Promedio de {materia}: {promedio:.1f}")

# Diccionario con listas como valores
horarios = {"lunes": ["9:00", "14:00"], "martes": ["10:00"]}
for dia, horas in horarios.items():
    print(dia, "->", len(horas), "citas")
```

## Unidad 16: Tipo de Dato Conjunto (set) y Función zip()

### 16.1 Definición, Unicidad y Métodos: set(), add(), remove()

**Problema que resuelve:** eliminar duplicados y comprobar pertenencia de forma muy rápida.

- Un `set` es una colección **mutable**, **sin orden** y **sin elementos repetidos**. Se crea con `{1, 2, 3}` o `set(secuencia)`.
- `{}` crea un diccionario vacío; para un conjunto vacío se usa `set()`.
- `add(x)` agrega un elemento (si ya existe, no pasa nada). `remove(x)` lo elimina y lanza `KeyError` si no existe; `discard(x)` lo elimina sin error.
- Los elementos deben ser inmutables (números, cadenas, tuplas). No se puede indexar un conjunto: `s[0]` falla.
- Convertir una lista a `set` elimina duplicados pero **pierde el orden**; `sorted(s)` devuelve una lista ordenada.

```python
# Ejemplo 1 (General): conjuntos
numeros = [3, 1, 2, 3, 2, 1, 5]
unicos = set(numeros)
print("Desduplicado:", unicos)
print("Ordenado:", sorted(unicos))
print("Cantidad de únicos:", len(unicos))

vacio = set()
print(type(vacio).__name__, "vs", type({}).__name__)

unicos.add(10)
unicos.add(3)          # ya existe: no cambia nada
print("Tras add:", sorted(unicos))

unicos.remove(10)
unicos.discard(99)     # no existe: sin error
print("Tras remove y discard:", sorted(unicos))

try:
    unicos.remove(99)
except KeyError as error:
    print("KeyError esperado en remove:", error)

print("3 in unicos:", 3 in unicos)

try:
    print(unicos[0])
except TypeError as error:
    print("Los conjuntos no se indexan:", error)
```

### 16.2 Operaciones Matemáticas entre Conjuntos

**Problema que resuelve:** comparar grupos completos (quién está en uno, en otro o en ambos) en una sola expresión.

| Operación | Operador | Método | Resultado |
|---|---|---|---|
| Unión | `a \| b` | `a.union(b)` | elementos de a **o** b |
| Intersección | `a & b` | `a.intersection(b)` | elementos en a **y** b |
| Diferencia | `a - b` | `a.difference(b)` | elementos de a que **no** están en b |
| Diferencia simétrica | `a ^ b` | `a.symmetric_difference(b)` | elementos en uno solo de los dos |

También existen `a <= b` (subconjunto) y `a >= b` (superconjunto). Observa que `a - b` **no** es lo mismo que `b - a`.

```python
# Ejemplo 1 (General): operaciones entre conjuntos
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Unión (a | b):", sorted(a | b))
print("Intersección (a & b):", sorted(a & b))
print("Diferencia (a - b):", sorted(a - b))
print("Diferencia (b - a):", sorted(b - a))
print("Diferencia simétrica (a ^ b):", sorted(a ^ b))

print("{1, 2} es subconjunto de a?", {1, 2} <= a)
print("a es superconjunto de {1, 2}?", a >= {1, 2})
print("a y {8, 9} no comparten nada?", a.isdisjoint({8, 9}))
```

### 16.3 Agregación y Desempaquetado: zip() y *

**Problema que resuelve:** recorrer varias listas paralelas a la vez y, cuando hace falta, separarlas de nuevo.

- `zip(a, b, c)` empareja el elemento 1 de cada secuencia, luego el 2, y así sucesivamente. Se detiene en la secuencia **más corta**.
- `zip` devuelve un iterador: se consume con un bucle o con `list(...)`.
- `dict(zip(claves, valores))` construye un diccionario desde dos listas.
- Desempaquetado inverso: `a, b = zip(*pares)`. El `*` expande la lista de pares como argumentos separados, y `zip` los reagrupa por columnas.

```python
# Ejemplo 1 (General): zip y desempaquetado con *
nombres = ["Ana", "Luis", "Maria"]
edades = [28, 35, 31]
ciudades = ["Panamá", "Colón", "David"]

for nombre, edad, ciudad in zip(nombres, edades, ciudades):
    print(f"{nombre} tiene {edad} años y vive en {ciudad}")

pares = list(zip(nombres, edades))
print("Pares:", pares)

diccionario = dict(zip(nombres, edades))
print("Diccionario:", diccionario)

# Desempaquetado inverso con *
nombres_2, edades_2 = zip(*pares)
print("Nombres recuperados:", nombres_2)
print("Edades recuperadas:", edades_2)

# zip se detiene en la secuencia más corta
print(list(zip([1, 2, 3, 4], ["a", "b"])))

# Un iterador de zip se consume una sola vez
z = zip([1, 2], [3, 4])
print(list(z), list(z))
```
